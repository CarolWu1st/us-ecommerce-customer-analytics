### Timestamp Data Quality Checks

Several timestamp inconsistencies were identified in the `orders` table:

- **1,359 orders** have `order_delivered_carrier_date` earlier than `order_approved_at`.
- **23 orders** have `order_delivered_customer_date` earlier than `order_delivered_carrier_date`.
- **8 delivered orders** are missing `order_delivered_customer_date`.

These records were not removed because they may still be valid for order-level analyses such as order volume, customer behavior, or revenue analysis.

Instead, three data quality flags were created:

- `flag_carrier_before_approval`
- `flag_delivered_before_carrier`
- `flag_delivered_missing_date`

These flags will be used to identify and exclude problematic records only in analyses where timestamp consistency is required, such as delivery-time and logistics performance calculations.

```python
#DQS

customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
order_items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
payments = pd.read_csv("../data/raw/olist_order_payments_dataset.csv")
reviews = pd.read_csv("../data/raw/olist_order_reviews_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
sellers = pd.read_csv("../data/raw/olist_sellers_dataset.csv")
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")
category_translation = pd.read_csv(
    "../data/raw/product_category_name_translation.csv"
)

tables = {
    "orders": orders,
    "customers": customers,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "geolocation": geolocation,
    "category_translation": category_translation
}

print("Number of tables loaded:", len(tables))
print(list(tables.keys()))
```

```python
dqs = []

for name, df in tables.items():
    total_cells = df.shape[0] * df.shape[1]
    missing_cells = df.isnull().sum().sum()

    dqs.append({
        "table": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "duplicate_rows": df.duplicated().sum(),
        "missing_cells": missing_cells,
        "missing_pct": round(missing_cells / total_cells * 100, 2)
    })

dqs_df = pd.DataFrame(dqs)

dqs_df
```

```python
def missing_summary(df):
    summary = pd.DataFrame({
        "missing_count": df.isnull().sum(),
        "missing_pct": (df.isnull().mean() * 100).round(2)
    })

    return summary[summary["missing_count"] > 0].sort_values(
        "missing_pct",
        ascending=False
    )


print("REVIEWS")
display(missing_summary(reviews))

print("\nPRODUCTS")
display(missing_summary(products))
```

### Missing Value Assessment

#### Reviews
Missing values in `review_comment_title` and `review_comment_message` are treated as expected business-level missingness rather than data quality errors. Customers can submit ratings without providing written comments. Therefore, these records are retained without imputation.

Text-based analyses will use only reviews with non-null comment content.

#### Products
Missing values are concentrated in product metadata fields, including category, name length, description length, and photo quantity. These records require further investigation before deciding whether to retain, label, or exclude them from category-level analyses.

A small number of products also have missing physical dimensions and will be examined separately.

### Review Table Key Structure

The `reviews` table contains 99,224 rows.

- `review_id` is not unique.
- `order_id` is not unique.
- The combination of `review_id` and `order_id` is unique across all rows.

Therefore, `review_id + order_id` is treated as the composite key for the review table.

Because some orders are associated with multiple review records, the relationship between `orders` and `reviews` should not be assumed to be strictly one-to-one. Care is required when joining review data with order-level metrics to avoid duplicate counting.

### Payment Data Quality Handling

The payment table contains a small number of unusual records:

- 9 payment records have a `payment_value` of 0.
- 6 of these are voucher records associated with orders that also contain other valid payment records.
- 3 zero-value records have `payment_type = not_defined`. These correspond to canceled orders with no order items, no approval timestamp, and no delivery activity, so they are treated as valid non-completed transaction records.
- 2 credit-card payment records have `payment_installments = 0`, which is not a valid installment count. Both records have positive payment values.

All payment records were retained.

Two data quality flags were created:

- `flag_zero_payment`
- `flag_invalid_installments`

The original `payment_installments` field was preserved. A new `payment_installments_clean` field was created, with invalid zero-installment values treated as missing rather than imputed.

## Data Cleaning Summary

The initial data quality assessment and cleaning process has been completed.

Key actions included:

- Converted order timestamp fields to datetime format.
- Identified and flagged timestamp inconsistencies in the order lifecycle.
- Confirmed uniqueness of primary and composite keys across core tables.
- Verified referential integrity between orders, customers, products, sellers, payments, reviews, and order items.
- Retained expected missing review text without imputation.
- Labeled missing product categories as `unknown` while preserving the original field.

- Preserved original values whenever the true replacement value could not be reliably inferred.

The cleaned datasets were exported to `data/processed/` for downstream exploratory data analysis.
